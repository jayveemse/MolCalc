## Decomposition reaction

$$\text{Li}_3\text{BN}_2(\text{s/l}) \rightleftharpoons \text{BN}(\text{s}) + 3\,\text{Li}(\text{g/l}) + \tfrac{1}{2}\,\text{N}_2(\text{g})$$

The sealed isochoric TZM autoclave yields the asymptotic equilibrium pressure $P_\text{eq}$ at each isothermal hold via the exponential fit $P(t) = P_\text{eq} + (P_0 - P_\text{eq})e^{-kt}$.

| Phase | State | T range | Notes |
|-------|-------|---------|-------|
| α-Li₃BN₂ | Solid | < 838 °C | Primary phase; tetragonal |
| Li₃BN₂ (l) | Liquid | > 863–916 °C | Melting point literature range |
| BN (hBN) | Solid | All T | Pure solid, $a = 1$ |
| Li₃N (l) | Liquid | > 813 °C | Product in Regime III |
| Li (l) | Liquid | > 181 °C | Condenses on autoclave walls |

---

## Equilibrium constant and Li vapour mode

With BN as a pure solid ($a = 1$) and liquid Li present (Regimes II/III):

$$K_1(T) = \left(\frac{\varphi_{N_2}\,P_{N_2}}{P^\circ}\right)^{1/2} \cdot a(\text{Li,liq})^3$$

| Mode | $K$ expression | When to use |
|------|---------------|-------------|
| **none** | $K = (P_\text{eq}/P^\circ)^{1/2}$ | Quick approximation |
| **saturated** *(recommended)* | $K = ((P_\text{eq} - P^\text{sat}_\text{Li})/P^\circ)^{1/2}$ | Li condenses; dominant regime |
| **stoichiometric** | $K = (P/7P^\circ)^{1/2}\cdot(6P/7P^\circ)^3$ | All Li gaseous; bounding case |

Li saturation pressure: $P^\text{sat}_\text{Li}(T)\;[\text{MPa}] = \exp(13.0699 - 8345.57/T)\times 10^{-6}$

---

## Phase rule and regime taxonomy

| Regime | Phases | $F$ | Consequence |
|--------|--------|-----|-------------|
| **II** | Li₃BN₂(l) + BN(s) + Li(l) + vapor | 1 | $P_\text{eq}$ set by $K_{1\ell}(T)$; dominant regime |
| **III** | Li₃BN₂(l) + BN(s) + Li₃N(l) + vapor | 1 | $P_\text{eq}$ set by $K_{4\ell}(T) > K_{1\ell}(T)$ |
| **INV** | all five phases | 0 | Invariant point $(T^*, P^*)$; intersection of Li₃BN₂ and Li₃N curves |

### Slope-kink: Regime II → III
The kink near ~950 °C marks the transition. Slope difference directly gives:
$$\Delta H^\circ_3 = \Delta H^\circ_{4\ell} - \Delta H^\circ_{1\ell} \approx +117.5 \pm 17.5\;\text{kJ/mol}\quad\text{(McHale et al. 1999)}$$

---

## Van't Hoff and Clausius-Clapeyron  (figures 3 and 4)

$$\ln K = -\frac{\Delta H^\circ}{R}\cdot\frac{1}{T} + \frac{\Delta S^\circ}{R} \qquad \log_{10} P_\text{eq} = -\frac{\Delta H^\circ}{R\ln 10}\cdot\frac{1}{T} + \text{const}$$

Slope $= -\Delta H^\circ/R$, intercept $= \Delta S^\circ/R$, zero-crossing $T_\text{eq} = \Delta H^\circ/\Delta S^\circ$.

---

## Free energy, Li mode comparison, and P–T diagram  (figures 6 and 7)

$$\Delta G^\circ(T) = \Delta H^\circ - T\,\Delta S^\circ \qquad \Delta G = \Delta G^\circ + RT\ln\!\left(\frac{P_\text{sys}}{P_\text{eq}}\right)$$

Operating with $P_\text{sys} < P_\text{eq}$ forces $\Delta G < 0$, removing Li and driving BN supersaturation.

Figure 7 shows the measured Li₃BN₂ equilibrium curve alongside the Li₃N equilibrium from Metzger et al. (2026) — the invariant point $(T^*, P^*)$ is where they intersect. The **K₃ cross-check** (figure 7b) plots $K_3 = K_1/K_4 = P_\text{eq}(\text{Li}_3\text{BN}_2)/P_\text{eq}(\text{Li}_3\text{N})$ vs. $1/T$; van't Hoff analysis should recover $\Delta H^\circ_3 \approx +117.5$ kJ/mol.

| Reference boundary | Value | Source |
|--------------------|-------|--------|
| $T_m(\text{Li}_3\text{BN}_2)$ | 863–916 °C | Maestre (2024) / Yamane (1987) |
| $T_m(\text{Li}_3\text{N})$ | 813 °C | Sangster & Pelton (1992) |
| Decomp. suppressed above | ~50 MPa | Maestre et al. (2024) |

```python
import numpy as np

# ════════════════════════════════════════════════════════════════════════
#  COLOURS AND THEME
# ════════════════════════════════════════════════════════════════════════
C_BLUE   = '#185FA5'
C_RED    = '#E24B4A'
C_GREEN  = '#1D9E75'
C_ORANGE = '#BA7517'
C_GREY   = '#888780'
C_BAND   = 'rgba(24,95,165,0.12)'

THEME = dict(
    template='simple_white',
    font=dict(family='Arial, sans-serif', size=12),
    margin=dict(l=60, r=20, t=55, b=55),
    legend=dict(bgcolor='rgba(255,255,255,0.8)', bordercolor='#ddd', borderwidth=1),
)

# ════════════════════════════════════════════════════════════════════════
#  WIDGET HELPERS
# ════════════════════════════════════════════════════════════════════════
W_INPUT  = w.Layout(width='305px')
W_SHORT  = w.Layout(width='130px')
W_XSHORT = w.Layout(width='95px')
W_ROW    = w.Layout(display='flex', flex_flow='row', align_items='center', margin='3px 0')
W_TAB    = w.Layout(padding='12px 16px', min_height='390px')
S0       = {'description_width': '0px'}

def section(text):
    return w.HTML(f'<div style="font-size:14px;font-weight:bold;color:{C_BLUE};'
                  f'margin:8px 0 4px;border-bottom:2px solid {C_BLUE};padding-bottom:2px">{text}</div>')

def small(text, colour='#666'):
    return w.HTML(f'<span style="font-size:12px;color:{colour}">{text}</span>')

def sep():
    return w.HTML('<hr style="border:none;border-top:1px solid #E0E0E0;margin:10px 0">')

def field_row(label_text, widget, hint_text=''):
    children = [
        w.HTML(f'<span style="width:255px;min-width:255px;display:inline-block;'
               f'font-size:13px;color:#333">{label_text}</span>'),
        widget,
    ]
    if hint_text:
        children.append(w.HTML(f'<span style="font-size:11px;color:#999;margin-left:6px">{hint_text}</span>'))
    return w.HBox(children, layout=W_ROW)

def browse_btn(label='Browse'):
    return w.Button(description=label,
                    style=w.ButtonStyle(button_color='#E8EDF5', text_color=C_BLUE),
                    layout=w.Layout(width='80px', height='28px'))


# ════════════════════════════════════════════════════════════════════════
#  FILE DIALOGS (tkinter, local Jupyter only)
# ════════════════════════════════════════════════════════════════════════
def _tk_open_file(filetypes=(('CSV','*.csv'),('All','*.*'))):
    try:
        import tkinter as tk; from tkinter import filedialog
        root = tk.Tk(); root.withdraw()
        try: root.attributes('-topmost', True)
        except Exception: pass
        p = filedialog.askopenfilename(filetypes=filetypes)
        root.destroy(); return p or ''
    except Exception as e: return f'[dialog error: {e}]'

def _tk_open_dir(initialdir='~'):
    try:
        import tkinter as tk; from tkinter import filedialog
        root = tk.Tk(); root.withdraw()
        try: root.attributes('-topmost', True)
        except Exception: pass
        p = filedialog.askdirectory(initialdir=os.path.expanduser(initialdir))
        root.destroy(); return p or ''
    except Exception as e: return f'[dialog error: {e}]'


# ════════════════════════════════════════════════════════════════════════
#  FORM — TAB 0: data & masses  (multi-run)
# ════════════════════════════════════════════════════════════════════════

# Colour palette — up to 8 simultaneous runs
RUN_PALETTE = [
    '#185FA5','#E24B4A','#1D9E75','#BA7517',
    '#7B2D8B','#2E7D5F','#A04000','#1565C0',
]

v_out_dir   = w.Text(value=os.path.expanduser('~'), layout=W_INPUT, style=S0)
v_li3n_file = w.Text(placeholder='optional Li3N companion CSV', layout=W_INPUT, style=S0)
v_mass_loss = w.Dropdown(
    options=[('Li and N2 leave crucible  (MW_loss = 34.83 g/mol)', 'Li_and_N2'),
             ('N2 only  (Li stays as liquid;  MW_loss = 14.01 g/mol)', 'N2_only')],
    value='Li_and_N2', layout=w.Layout(width='400px'), style=S0)

btn_br_li3n = browse_btn()
btn_br_out  = browse_btn('Folder')
btn_br_li3n.on_click(lambda b: v_li3n_file.__setattr__('value', _tk_open_file() or v_li3n_file.value))
btn_br_out.on_click( lambda b: v_out_dir.__setattr__('value',   _tk_open_dir(v_out_dir.value) or v_out_dir.value))

# ── dynamic run list ───────────────────────────────────────────────────────
# Each entry is a dict: {file, label, m_initial, m_final, color_idx, widgets}
_run_entries = []   # list of dicts

def _color_swatch(hex_col):
    return w.HTML(
        f'<div style="width:14px;height:14px;border-radius:50%;'
        f'background:{hex_col};display:inline-block;margin:0 4px;'
        f'border:1px solid rgba(0,0,0,0.2)"></div>')

def _make_run_row(idx):
    col = RUN_PALETTE[idx % len(RUN_PALETTE)]
    swatch   = _color_swatch(col)
    f_text   = w.Text(placeholder='CSV path', layout=w.Layout(width='240px'), style=S0)
    f_btn    = browse_btn()
    lbl_text = w.Text(value=f'Run {idx+1}', placeholder='label',
                      layout=w.Layout(width='80px'), style=S0)
    mi_text  = w.Text(value='0.7500', placeholder='m_i (g)',
                      layout=w.Layout(width='72px'), style=S0)
    mf_text  = w.Text(value='', placeholder='m_f (g)',
                      layout=w.Layout(width='72px'), style=S0)
    rm_btn   = w.Button(description='✕',
                        style=w.ButtonStyle(button_color='#FCDCD9', text_color='#C0392B'),
                        layout=w.Layout(width='32px', height='28px'))

    row_box = w.HBox(
        [swatch, f_text, f_btn, lbl_text,
         w.HTML('<span style="font-size:11px;color:#999;margin:0 3px">m_i</span>'),
         mi_text,
         w.HTML('<span style="font-size:11px;color:#999;margin:0 3px">m_f</span>'),
         mf_text, rm_btn],
        layout=w.Layout(align_items='center', margin='2px 0', gap='4px'))

    entry = dict(file=f_text, label=lbl_text, m_initial=mi_text, m_final=mf_text,
                 color=col, row=row_box)

    def _on_browse(b, e=entry):
        p = _tk_open_file()
        if p: e['file'].value = p
    def _on_remove(b, e=entry):
        _run_entries.remove(e)
        _refresh_run_list()

    f_btn.on_click(_on_browse)
    rm_btn.on_click(_on_remove)
    return entry

runs_vbox = w.VBox([], layout=w.Layout(margin='4px 0'))

def _refresh_run_list():
    runs_vbox.children = tuple(e['row'] for e in _run_entries)

def _add_run(b=None):
    if len(_run_entries) >= 8:
        return
    entry = _make_run_row(len(_run_entries))
    _run_entries.append(entry)
    _refresh_run_list()

btn_add_run = w.Button(
    description='+ Add run',
    style=w.ButtonStyle(button_color='#E8EDF5', text_color=C_BLUE),
    layout=w.Layout(width='100px', height='30px'))
btn_add_run.on_click(_add_run)

# start with one empty run
_add_run()

tab0 = w.VBox([
    section('Runs'),
    w.HTML('<span style="font-size:12px;color:#666">'
           'Add one CSV per run. Each gets its own colour on shared figures. '
           'Label, m_i, and m_f are per-run.</span>'),
    w.HBox([w.HTML(
        '<span style="font-size:11px;color:#888;width:20px;display:inline-block"></span>'
        '<span style="font-size:11px;color:#888;width:244px;display:inline-block">CSV file</span>'
        '<span style="font-size:11px;color:#888;width:88px;display:inline-block">Label</span>'
        '<span style="font-size:11px;color:#888;width:76px;display:inline-block">m_initial (g)</span>'
        '<span style="font-size:11px;color:#888;width:76px;display:inline-block">m_final (g)</span>'
    )], layout=w.Layout(margin='6px 0 2px 0')),
    runs_vbox,
    w.HBox([btn_add_run], layout=w.Layout(margin='6px 0')),

    sep(), section('What leaves the crucible?'),
    field_row('Mass loss mode', v_mass_loss),

    sep(), section('Li3N reference  (Hess K3 cross-check)'),
    w.HTML('<span style="font-size:12px;color:#666">Optional companion run — shared across all runs.</span>'),
    w.HBox([v_li3n_file, btn_br_li3n], layout=w.Layout(margin='4px 0 8px', gap='6px')),

    sep(), section('Output folder'),
    w.HTML('<span style="font-size:12px;color:#666">Plots and CSVs are saved here.</span>'),
    w.HBox([v_out_dir, btn_br_out], layout=w.Layout(margin='4px 0', gap='6px')),
], layout=W_TAB)

# ── TAB 1: analysis modes ──────────────────────────────────────────────────
v_li_mode = w.RadioButtons(
    options=[
        ('none  --  K = (P / P0)^0.5   [ignore Li(g)]',                   'none'),
        ('saturated  --  K = ((P - P_Li_sat) / P0)^0.5   [recommended]',  'saturated'),
        ('stoichiometric  --  K = (P/7P0)^0.5 * (6P/7P0)^3  [bounding]',  'stoichiometric'),
    ], value='saturated', layout=w.Layout(margin='4px 0 8px 8px'))
v_t_kink  = w.Text(value='', placeholder='blank = auto', layout=W_SHORT, style=S0)
v_dH_li3n = w.Text(value='290.0', layout=W_XSHORT, style=S0)
v_dS_li3n = w.Text(value='267.0', layout=W_XSHORT, style=S0)

tab1 = w.VBox([
    section('Li vapour mode'),
    small('"saturated" is correct for most runs -- Li condenses on the autoclave walls.'),
    v_li_mode, sep(),
    section('Slope-kink analysis  (Regime II to III transition)'),
    field_row('Breakpoint T  (degC)', v_t_kink, 'Auto-detect via piecewise SSR if blank'),
    small('Slope difference should recover dH3 ~ +117.5 +/- 17.5 kJ/mol (McHale 1999).', '#888'),
    sep(), section('Li3N reference data  (Hess K3 cross-check)'),
    small('Optional companion run. Enables K3 = K1/K4 Hess cross-check and phase diagram overlay.'),
    w.HBox([v_li3n_file, btn_br_li3n], layout=w.Layout(margin='4px 0 8px', gap='6px')),
    field_row('Li3N  dH  (kJ/mol)', v_dH_li3n, 'Phase diagram overlay  -- Metzger et al. 2026'),
    field_row('Li3N  dS  (J/mol/K)', v_dS_li3n, '267 J/(mol K)  -- Metzger et al. 2026'),
], layout=W_TAB)

# ── TAB 2: column mapping ──────────────────────────────────────────────────
COL_DEFS = [
    ('T_K  -- equilibrium temperature (K)',  'Control_Temp_at_End_K',  'Derived from T_C if absent -- leave blank to auto-derive'),
    ('T_C  -- temperature (degC)',           'Control_Temp_at_End_C',  'Same in Celsius'),
    ('P_eq  -- equilibrium pressure (MPa)', 'Asymptote_MPa',           'Asymptotic pressure from fit'),
    ('n_gas  -- cumulative N2 (mmol)',       'n_eq_mmol',               'Auto-derived as cumsum(dn) if absent -- leave blank to auto-derive'),
    ('dn  -- incremental gas (mmol)',        'Delta_n_mmol',            'Gas released this step'),
    ('sigma_dn  -- uncertainty on dn',      'Sigma_delta_n_mmol',      '1-sigma std dev'),
    ('LOD  -- 3-sigma limit of detection',  'LOD_3sigma_mmol',         'Steps below this are flagged'),
]
COL_KEYS    = ['T_K','T_C','P_eq_MPa','n_gas_mmol','dn_mmol','sigma_dn','LOD']
col_widgets = [w.Text(value=default, layout=W_INPUT, style=S0) for _,default,_ in COL_DEFS]

tab2 = w.VBox([
    section('CSV column names'),
    small('Must match your CSV headers exactly (case-sensitive).'), sep(),
] + [field_row(lbl, wgt, hint) for (lbl,_,hint),wgt in zip(COL_DEFS, col_widgets)],
    layout=W_TAB)

# ── TAB 3: advanced ────────────────────────────────────────────────────────
v_p0 = w.Text(value='0.1',   layout=W_XSHORT, style=S0)
v_R  = w.Text(value='8.314', layout=W_XSHORT, style=S0)
v_outlier = w.Text(value='3.5', layout=W_XSHORT, style=S0)

tab3 = w.VBox([
    section('Standard state'),
    field_row('P0  (MPa)',       v_p0, '1 bar = 0.1 MPa  (IUPAC)'),
    field_row('R  (J/mol/K)',    v_R,  '8.314 J/(mol K)'),
    sep(), section('Outlier detection'),
    field_row('Modified Z-score threshold', v_outlier, 'Iglewicz & Hoaglin -- standard: 3.5'),
    sep(), section('Reference MW  (g/mol)'),
    w.HTML('<table style="font-family:monospace;font-size:12px;border-collapse:collapse;margin:4px 0">'
           '<tr style="background:#EEF2FF"><th style="padding:4px 12px">Species</th>'
           '<th style="padding:4px 12px;text-align:right">MW</th></tr>'
           '<tr><td style="padding:3px 12px">Li3BN2</td><td style="padding:3px 12px;text-align:right">59.648</td></tr>'
           '<tr><td style="padding:3px 12px">BN</td>    <td style="padding:3px 12px;text-align:right">24.818</td></tr>'
           '<tr><td style="padding:3px 12px">Li3N</td>  <td style="padding:3px 12px;text-align:right">34.830</td></tr>'
           '<tr><td style="padding:3px 12px">Li</td>    <td style="padding:3px 12px;text-align:right">6.941</td></tr>'
           '<tr><td style="padding:3px 12px">N2</td>    <td style="padding:3px 12px;text-align:right">28.014</td></tr>'
           '</table>'),
], layout=W_TAB)

# ════════════════════════════════════════════════════════════════════════
#  ASSEMBLE UI  (form only — plots rendered below separately)
# ════════════════════════════════════════════════════════════════════════
tabs = w.Tab(children=[tab0,tab1,tab2,tab3], layout=w.Layout(width='700px'))
for i,t in enumerate(['Data & masses','Analysis modes','Column mapping','Advanced']):
    tabs.set_title(i,t)

header = w.HTML(
    f'<div style="background:{C_BLUE};color:white;padding:12px 18px;border-radius:6px 6px 0 0">'
    '<span style="font-size:17px;font-weight:bold">Li3BN2 decomposition thermodynamics</span><br>'
    '<span style="font-size:12px;color:#C8DEFF">Li3BN2(s/l) = BN(s) + 3 Li(g/l) + 1/2 N2(g)</span></div>')

btn_run   = w.Button(description='Run analysis',
                     style=w.ButtonStyle(button_color=C_BLUE,    text_color='white'),
                     layout=w.Layout(width='140px', height='36px'))
btn_save  = w.Button(description='Save plots',
                     style=w.ButtonStyle(button_color=C_GREEN,   text_color='white'),
                     layout=w.Layout(width='140px', height='36px'), disabled=True)
btn_clear = w.Button(description='Clear plots',
                     style=w.ButtonStyle(button_color='#E0E0E0', text_color='#444'),
                     layout=w.Layout(width='110px', height='36px'))
status = w.HTML('<span style="font-size:12px;color:#888">Ready.</span>')
btn_row = w.HBox([btn_run,btn_save,btn_clear,status],
                 layout=w.Layout(margin='8px 0 4px', align_items='center', gap='8px'))

# text log — scrollable, fixed height so it never pushes plots off screen
out_text = w.Output(layout=w.Layout(
    width='700px',
    height='180px',        # fixed height: always exactly this tall
    max_height='180px',
    overflow_y='scroll',   # scroll bar always visible when content overflows
    border='1px solid #E0E0E0', border_radius='4px',
    padding='8px', font_family='monospace', font_size='12px',
    background_color='#FAFAFA'))

# per-plot save buttons — one button row added per figure, pure widgets
# kept completely separate from out_plots so HTML rendering is never disturbed
out_btns = w.Output(layout=w.Layout(width='700px', margin='4px 0'))

# plot output — PURE HTML only, no widgets ever go in here
out_plots = w.Output(layout=w.Layout(width='100%', margin='6px 0'))

display(w.VBox([header, tabs, btn_row, out_btns, out_plots, out_text]))


# ════════════════════════════════════════════════════════════════════════
#  FIGURE STORE  (go.Figure objects for saving)
# ════════════════════════════════════════════════════════════════════════
_figures = []   # list of (go.Figure, base_name_str)


# ════════════════════════════════════════════════════════════════════════
#  LOG CAPTURE
# ════════════════════════════════════════════════════════════════════════
class _Tee:
    """Redirect sys.stdout to both a log buffer and the out_text widget (thread-safe)."""
    def __init__(self, orig):
        self._buf = io.StringIO(); self._orig = orig
    def write(self, t):
        if not t: return
        self._buf.write(t)
        try: out_text.append_stdout(t)   # thread-safe widget display
        except Exception: pass
    def flush(self):
        self._buf.flush()
    def getvalue(self): return self._buf.getvalue()


# ════════════════════════════════════════════════════════════════════════
#  CONFIG
# ════════════════════════════════════════════════════════════════════════
def collect_config():
    def _f(wgt, default=None):
        s = wgt.value.strip()
        if not s: return default
        try: return float(s)
        except ValueError: return default

    runs = []
    for e in _run_entries:
        fp = e['file'].value.strip()
        if fp:
            runs.append(dict(
                DATA_FILE   = fp,
                LABEL       = e['label'].value.strip() or fp,
                COLOR       = e['color'],
                M_INITIAL_G = _f(e['m_initial']),
                M_FINAL_G   = _f(e['m_final']),
            ))

    return dict(
        RUNS             = runs,
        LI3N_FILE        = v_li3n_file.value.strip() or None,
        MASS_LOSS_IS     = v_mass_loss.value,
        LI_GAS_MODE      = v_li_mode.value,
        T_KINK_C         = (lambda s: float(s) if s else None)(v_t_kink.value.strip()),
        P0_MPA           = (lambda s: float(s) if s else 0.1)(v_p0.value.strip()),
        R_GAS            = (lambda s: float(s) if s else 8.314)(v_R.value.strip()),
        OUTLIER_THRESHOLD= (lambda s: float(s) if s else 3.5)(v_outlier.value.strip()),
        COL_KEYS         = COL_KEYS,
        COL_VALS         = [wgt.value.strip() for wgt in col_widgets],
        dH_Li3N_lit      = (lambda s: float(s) if s else 290.0)(v_dH_li3n.value.strip()),
        dS_Li3N_lit      = (lambda s: float(s) if s else 267.0)(v_dS_li3n.value.strip()),
        OUT_DIR          = v_out_dir.value.strip(),
    )


# ════════════════════════════════════════════════════════════════════════
#  PLOTLY HELPERS
# ════════════════════════════════════════════════════════════════════════
def _band(x, y, ci, color=C_BAND, name='95% CI'):
    return go.Scatter(
        x=np.concatenate([x, x[::-1]]),
        y=np.concatenate([y+ci, (y-ci)[::-1]]),
        fill='toself', fillcolor=color,
        line=dict(color='rgba(0,0,0,0)'),
        showlegend=True, name=name, legendgroup=name)

def _cscatter(x, y, c_vals, name='data', size=7, cbar_title='T (K)', show_cbar=True,
              customdata=None, htmpl=None, **kw):
    return go.Scatter(
        x=x, y=y, mode='markers', name=name,
        marker=dict(
            color=c_vals, colorscale='plasma', size=size,
            showscale=show_cbar,
            colorbar=dict(title=cbar_title, thickness=14, len=0.8) if show_cbar else None,
            line=dict(color='rgba(0,0,0,0.3)', width=0.5)),
        customdata=customdata,
        hovertemplate=htmpl, **kw)

def _annot(fig, text, x=0.98, y=0.05, xa='right', ya='bottom',
           xref='paper', yref='paper'):
    fig.add_annotation(
        x=x, y=y, xref=xref, yref=yref,
        text=text.replace('\n','<br>'),
        showarrow=False, align='left' if xa=='left' else 'right',
        xanchor=xa, yanchor=ya,
        bordercolor='#ccc', borderwidth=1, borderpad=6,
        bgcolor='rgba(255,255,255,0.88)',
        font=dict(size=10, family='monospace, Courier'))

def _show(fig, base_name, title_str):
    """
    Render figure as pure HTML (no widgets in out_plots).
    Add a per-plot Save PNG button to out_btns (completely separate widget area).
    Both operations are thread-safe via append_display_data / append_stdout.
    """
    fig_idx = len(_figures) + 1
    div_id  = f'plydiv_{fig_idx}'

    # ── HTML figure (pure HTML, no widget protocol) ───────────────────────
    html_str = fig.to_html(
        include_plotlyjs='cdn', full_html=False,
        div_id=div_id,
        config={
            'displayModeBar': True,
            'scrollZoom': True,
            'toImageButtonOptions': {'format': 'png', 'scale': 2,
                                     'filename': base_name},
        })

    fig_heading = (
        f'<div style="font-size:13px;font-weight:bold;color:{C_BLUE};'
        f'margin:14px 0 2px">{fig_idx}. {title_str}</div>'
        f'<hr style="border:none;border-top:1px solid #eee;margin:2px 0 6px">'
    )
    # pure HTML → goes into out_plots with no widget mixing
    out_plots.append_display_data(HTML(fig_heading + html_str))

    # ── per-plot save button (widget) → goes into out_btns only ──────────
    def _make_btn(fig_ref, name_ref, idx_ref, title_ref):
        lbl = w.HTML(
            f'<span style="font-size:12px;color:#555;min-width:220px;'
            f'display:inline-block">{idx_ref}. {title_ref}</span>')
        btn = w.Button(
            description='Save PNG',
            style=w.ButtonStyle(button_color=C_GREEN, text_color='white'),
            layout=w.Layout(width='100px', height='28px'))
        status = w.HTML(
            '<span style="font-size:11px;color:#aaa;margin-left:8px"> </span>')

        def _on_click(b):
            OUT = v_out_dir.value.strip()
            os.makedirs(OUT, exist_ok=True)
            ts    = datetime.now().strftime('%y%m%d_%H%M%S')
            fname = f'{name_ref}_{ts}.png'
            fpath = os.path.join(OUT, fname)
            try:
                fig_ref.write_image(fpath, scale=2)
                status.value = (
                    f'<span style="font-size:11px;color:{C_GREEN}">'
                    f' Saved: {fname}</span>')
            except Exception as exc:
                status.value = (
                    f'<span style="font-size:11px;color:{C_RED}">'
                    f' Error: {exc}</span>')
        btn.on_click(_on_click)
        return w.HBox([lbl, btn, status],
                      layout=w.Layout(align_items='center',
                                      margin='2px 0', padding='2px 6px'))

    btn_row_fig = _make_btn(fig, base_name, fig_idx, title_str)
    out_btns.append_display_data(btn_row_fig)

    _figures.append((fig, base_name))
    out_text.append_stdout(f'  Plot {fig_idx} ready: {title_str}\n')



def run_analysis(cfg):
    import pandas as pd
    from scipy import stats

    P0=cfg['P0_MPA']; R=cfg['R_GAS']; MODE=cfg['LI_GAS_MODE']; LOSS=cfg['MASS_LOSS_IS']
    NU_N2=0.5; NU_LI=3.0
    MW=dict(Li3BN2=3*6.941+10.811+2*14.007, BN=10.811+14.007,
            Li3N=3*6.941+14.007, Li=6.941, N2=28.014)
    MW_LOSS=(3*MW['Li']+NU_N2*MW['N2']) if LOSS=='Li_and_N2' else NU_N2*MW['N2']

    RUNS = cfg['RUNS']
    if not RUNS:
        raise ValueError('No runs configured — add at least one CSV file.')

    def P_Li_sat(T_K): return np.exp(13.0699-8345.57/np.asarray(T_K,float))/1e6

    def compute_ln_K(P_eq, T_K):
        P_eq=np.asarray(P_eq,float); T_K=np.asarray(T_K,float)
        if MODE=='none':
            return 0.5*np.log(P_eq/P0), P_eq, np.zeros_like(P_eq)
        elif MODE=='saturated':
            P_Li=P_Li_sat(T_K); P_N2=np.maximum(P_eq-P_Li,1e-12)
            return 0.5*np.log(P_N2/P0), P_N2, P_Li
        else:
            P_N2=P_eq/7.0; P_Li=6.0*P_N2
            return (0.5*np.log(P_N2/P0)+3.0*np.log(np.maximum(P_Li/P0,1e-30))), P_N2, P_Li

    def vhfit(iT, lnK, wts=None):
        if wts is None: sl,ic,rv,_,_=stats.linregress(iT,lnK)
        else:
            ww=wts/wts.sum()
            Sw=ww.sum();Swx=(ww*iT).sum();Swy=(ww*lnK).sum()
            Swxx=(ww*iT**2).sum();Swxy=(ww*iT*lnK).sum()
            det=Sw*Swxx-Swx**2
            sl=(Sw*Swxy-Swx*Swy)/det; ic=(Swxx*Swy-Swx*Swxy)/det
            rv=np.corrcoef(iT,lnK)[0,1]
        return -sl*R/1e3, ic*R, -sl/ic, rv**2, ic, sl

    def piecewise(iT, lnK, T_bk=None):
        n=len(iT)
        # need at least 4 points to split into two segments of >=2 each
        if n < 4:
            sl,ic,rv,_,_=stats.linregress(iT,lnK)
            return dict(idx=n//2, T_bk=1/iT[n//2] if n>1 else 1/iT[0],
                        sl1=sl,ic1=ic,r2_1=rv**2,sl2=sl,ic2=ic,r2_2=rv**2,
                        dH1=-sl*R/1e3,dS1=ic*R,Teq1=-sl/ic if ic!=0 else 0,
                        dH2=-sl*R/1e3,dS2=ic*R,Teq2=-sl/ic if ic!=0 else 0,
                        dH3=0.0)
        def ssr(i):
            if i<2 or i>n-2: return np.inf
            s1,c1,*_=stats.linregress(iT[:i],lnK[:i])
            s2,c2,*_=stats.linregress(iT[i:],lnK[i:])
            return ((lnK[:i]-(s1*iT[:i]+c1))**2).sum()+((lnK[i:]-(s2*iT[i:]+c2))**2).sum()
        idx=(np.searchsorted(-iT,-1/T_bk) if T_bk else np.argmin([ssr(i) for i in range(n)]))
        idx=max(2,min(n-2,idx))
        s1,c1,r1,*_=stats.linregress(iT[:idx],lnK[:idx])
        s2,c2,r2,*_=stats.linregress(iT[idx:],lnK[idx:])
        return dict(idx=idx,T_bk=1/iT[idx],
                    sl1=s1,ic1=c1,r2_1=r1**2,sl2=s2,ic2=c2,r2_2=r2**2,
                    dH1=-s1*R/1e3,dS1=c1*R,Teq1=-s1/c1,
                    dH2=-s2*R/1e3,dS2=c2*R,Teq2=-s2/c2,
                    dH3=(-s2+s1)*R/1e3)

    T_bk_K=(cfg['T_KINK_C']+273.15) if cfg['T_KINK_C'] else None

    # ── load and process each run ─────────────────────────────────────────
    processed = []   # one dict per run, filled below

    for run_cfg in RUNS:
        lbl = run_cfg['LABEL']
        col = run_cfg['COLOR']
        print(f'\nLoading [{lbl}]: {run_cfg["DATA_FILE"]}')
        df_raw=pd.read_csv(run_cfg['DATA_FILE']); df_raw.columns=df_raw.columns.str.strip()
        print(f'  {len(df_raw)} rows')

        # ── flexible column resolution: try user mapping first,
        #    then a built-in list of known aliases for each internal key ──────
        ALIASES = {
            'T_K'        : ['Control_Temp_at_End_K', 'T_K', 'Temp_K',
                             'Temperature_K', 'T(K)', 'TK'],
            'T_C'        : ['Sample_Temp_at_End_C', 'Control_Temp_at_End_C',
                             'T_C', 'Temp_C', 'Temperature_C', 'T(C)', 'TC', 'Temp_degC'],
            'P_eq_MPa'   : ['Asymptote_MPa', 'P_eq_MPa', 'P_eq', 'P_MPa',
                             'Pressure_MPa', 'Peq_MPa'],
            'n_gas_mmol' : ['n_eq_mmol', 'n_gas_mmol', 'n_N2_mmol',
                             'ngas_mmol', 'N2_cumul_mmol'],
            'dn_mmol'    : ['Delta_n_mmol', 'dn_mmol', 'delta_n_mmol',
                             'dn', 'Dn_mmol'],
            'sigma_dn'   : ['Sigma_delta_n_mmol', 'sigma_dn', 'Sigma_dn',
                             'sigma_delta_n', 'err_dn_mmol'],
            'LOD'        : ['LOD_3sigma_mmol', 'LOD', 'lod_mmol',
                             'LOD_mmol', 'detection_limit_mmol'],
        }

        cols = list(df_raw.columns)
        rename = {}

        # 1. user-specified mapping (column mapping tab) takes priority
        for csv_h, int_k in zip(cfg['COL_VALS'], cfg['COL_KEYS']):
            if csv_h and csv_h in cols:
                rename[csv_h] = int_k

        # 2. fall back to known aliases for any internal key not yet mapped
        mapped_internal = set(rename.values())
        for int_k, candidates in ALIASES.items():
            if int_k not in mapped_internal:
                for cand in candidates:
                    if cand in cols and cand not in rename:
                        rename[cand] = int_k
                        break

        df = df_raw.rename(columns=rename).copy()

        # report what was resolved
        resolved   = {v:k for k,v in rename.items()}
        unresolved = [k for k in cfg['COL_KEYS'] if k not in df.columns]
        if resolved:
            print('  Column mapping: ' +
                  '  '.join(f'{v}->{k}' for k,v in resolved.items()))
        if unresolved:
            print(f'  Unresolved cols (will be derived or skipped): {unresolved}')

        # warn if falling back from Sample_Temp to Control_Temp for T_C
        t_c_source = resolved.get('T_C')
        if t_c_source == 'Control_Temp_at_End_C':
            print('  WARNING: Sample_Temp_at_End_C not found -- '
                  'using Control_Temp_at_End_C for temperature. '
                  'Verify this is the intended temperature column.')
        elif t_c_source == 'Sample_Temp_at_End_C':
            print('  Temperature source: Sample_Temp_at_End_C')

        # ── derive T_K / T_C if either is still missing ───────────────────
        if 'T_K' not in df.columns:
            if 'T_C' in df.columns:
                df['T_K'] = df['T_C'] + 273.15
            else:
                raise KeyError(
                    'Cannot find a temperature column. '
                    f'Available: {list(df.columns)}. '
                    'Check Column mapping tab or rename your headers.')
        if 'T_C' not in df.columns:
            df['T_C'] = df['T_K'] - 273.15

        # ── derive cumulative N2 if absent ────────────────────────────────
        if 'n_gas_mmol' not in df.columns:
            if 'dn_mmol' in df.columns:
                df['n_gas_mmol'] = df['dn_mmol'].cumsum()
            else:
                df['n_gas_mmol'] = 0.0

        # ── single-point mode: disable outlier/LOD filters ────────────────
        n_rows = len(df)
        single_point = (n_rows == 1)
        if single_point:
            print('  Single-point run -- skipping outlier/LOD filtering and fitting')

        df['inv_T']=1.0/df['T_K']
        df['log10_P']=np.log10(df['P_eq_MPa'])
        df['ln_Kp_raw']=0.5*np.log(df['P_eq_MPa']/P0)

        if single_point or 'sigma_dn' not in df.columns or 'LOD' not in df.columns:
            df['sigma_dn']   = df.get('sigma_dn',   pd.Series([0.001]*n_rows))
            df['LOD']        = df.get('LOD',         pd.Series([0.0]*n_rows))
            df['dn_mmol']    = df.get('dn_mmol',     pd.Series([0.0]*n_rows))
            df['SNR']        = 999.0
            df['modZ']       = 0.0
            df['Kp_outlier'] = False
            df['above_LOD']  = True
            df['use_in_fit'] = True
        else:
            df['SNR']=df['dn_mmol']/df['sigma_dn']
            med=df['ln_Kp_raw'].median(); mad=np.median(np.abs(df['ln_Kp_raw']-med))
            df['modZ']=0.6745*(df['ln_Kp_raw']-med)/(mad+1e-30)
            df['Kp_outlier']=df['modZ'].abs()>cfg['OUTLIER_THRESHOLD']
            df['above_LOD']=df['dn_mmol']>df['LOD']
            df['use_in_fit']=df['above_LOD']&~df['Kp_outlier']

        ok=df['use_in_fit']; bad=~ok
        if not single_point:
            print(f'  Used: {ok.sum()}  Outliers: {df["Kp_outlier"].sum()}  Below LOD: {(~df["above_LOD"]).sum()}')

        d=df[ok].copy().reset_index(drop=True)
        T_C_a=d['T_C'].values; T_K_a=d['T_K'].values
        iT_a=d['inv_T'].values; P_a=d['P_eq_MPa'].values; ng_a=d['n_gas_mmol'].values
        n_pts=len(d)

        ln_K_a, P_N2_a, P_Li_a = compute_ln_K(P_a, T_K_a)
        log10_a = np.log10(P_a)
        n_rxn_a = ng_a/NU_N2

        # single-point: skip all fitting, push a placeholder entry and move on
        if single_point or n_pts < 2:
            dH_u=dS_u=Teq_u=r2_u=dH_err=dS_err=Teq_err=0.0
            ic_u=sl_u=sl_log=ic_log=dH_log=r_log=0.0
            x_fit=iT_a.copy(); y_fit=ln_K_a.copy()
            se_b=np.zeros_like(x_fit); t95=2.0
            resid_u=np.zeros(n_pts)
            kink=dict(idx=0,T_bk=T_K_a[0],sl1=0,ic1=0,r2_1=0,
                      sl2=0,ic2=0,r2_2=0,dH1=0,dS1=0,Teq1=T_K_a[0],
                      dH2=0,dS2=0,Teq2=T_K_a[0],dH3=0)
            mass_ok=False; n_rxn_tot=conv=None; frac_d=None
            print(f'  Point: T={T_C_a[0]:.1f} C  P={P_a[0]:.4f} MPa  log10P={log10_a[0]:.3f}')
            processed.append(dict(
                lbl=lbl, col=col, df=df, d=d, ok=ok, bad=bad,
                T_C_a=T_C_a, T_K_a=T_K_a, iT_a=iT_a, P_a=P_a,
                ng_a=ng_a, n_rxn_a=n_rxn_a, frac_d=frac_d,
                log10_a=log10_a, ln_K_a=ln_K_a, P_N2_a=P_N2_a,
                P_Li_a=P_Li_a, dH_u=dH_u, dS_u=dS_u, Teq_u=Teq_u,
                r2_u=r2_u, dH_err=dH_err, dS_err=dS_err,
                Teq_err=Teq_err, ic_u=ic_u, sl_u=sl_u,
                sl_log=sl_log, ic_log=ic_log, dH_log=dH_log,
                r_log=r_log, x_fit=x_fit, y_fit=y_fit,
                se_b=se_b, t95=t95, resid_u=resid_u, kink=kink,
                mass_ok=mass_ok, n_rxn_tot=n_rxn_tot, conv=conv,
                n_pts=n_pts, single_point=True,
            ))
            continue

        dH_u,dS_u,Teq_u,r2_u,ic_u,sl_u = vhfit(iT_a, ln_K_a)
        wa=1.0/d['sigma_dn'].values**2
        Sw=wa.sum();Swx=(wa*iT_a).sum();Swxx=(wa*iT_a**2).sum()
        det=Sw*Swxx-Swx**2
        dH_err=np.sqrt(Sw/det)*R/1e3 if det>0 else 0
        dS_err=np.sqrt(Swxx/det)*R   if det>0 else 0
        Teq_err=(Teq_u*np.sqrt((np.sqrt(Sw/det)/sl_u)**2+(np.sqrt(Swxx/det)/ic_u)**2)
                 if det>0 and sl_u!=0 and ic_u!=0 else 0)

        sl_log,ic_log,r_log,*_=stats.linregress(iT_a,log10_a)
        dH_log=-sl_log*R*np.log(10)/1e3

        x_bar=iT_a.mean()
        x_fit=np.linspace(iT_a.min()*0.998, iT_a.max()*1.002, 300)
        y_fit=sl_u*x_fit+ic_u
        resid_u=ln_K_a-(sl_u*iT_a+ic_u)
        s_res=np.sqrt((resid_u**2).sum()/(n_pts-2)) if n_pts>2 else 0
        se_b=s_res*np.sqrt(1/n_pts+(x_fit-x_bar)**2/((iT_a-x_bar)**2).sum()) if n_pts>2 else np.zeros_like(x_fit)
        t95=stats.t.ppf(0.975, df=n_pts-2) if n_pts>2 else 2.0

        kink=piecewise(iT_a, ln_K_a, T_bk_K)

        # mass balance
        mass_ok = run_cfg['M_FINAL_G'] is not None and run_cfg['M_INITIAL_G'] is not None
        if mass_ok:
            dm=run_cfg['M_INITIAL_G']-run_cfg['M_FINAL_G']
            n_rxn_tot=dm/MW_LOSS*1000
            n0=run_cfg['M_INITIAL_G']/MW['Li3BN2']*1000
            conv=n_rxn_tot/n0*100
        else:
            n_rxn_tot=conv=None

        frac_d=(n_rxn_a/(run_cfg['M_INITIAL_G']/MW['Li3BN2']*1000)*100
                if run_cfg.get('M_INITIAL_G') else None)

        print(f'  dH = {dH_u:+.2f} +/- {dH_err:.2f} kJ/mol  '
              f'dS = {dS_u:+.2f} J/(mol K)  '
              f'T_eq = {Teq_u-273.15:.0f} C  R2={r2_u:.4f}')
        print(f'  Kink {kink["T_bk"]-273.15:.0f} C: dH1={kink["dH1"]:+.1f} dH4={kink["dH2"]:+.1f} dH3={kink["dH3"]:+.1f} kJ/mol')

        # save per-run results CSV
        OUT=cfg['OUT_DIR']; os.makedirs(OUT,exist_ok=True)
        exp=d[['T_K','T_C','P_eq_MPa','n_gas_mmol','dn_mmol','sigma_dn']].copy()
        exp['inv_T']=iT_a; exp['P_N2_MPa']=P_N2_a; exp['P_Li_MPa']=P_Li_a
        exp['ln_K']=ln_K_a; exp['log10_P']=log10_a; exp['n_rxn_mmol']=n_rxn_a
        exp['dG_kJ']=(dH_u*1e3-d['T_K'].values*dS_u)/1e3
        ts_now=datetime.now().strftime('%y%m%d_%H%M%S')
        safe_lbl=lbl.replace(' ','_').replace('/','_')
        csv_out=os.path.join(OUT,f'{safe_lbl}_thermo_{ts_now}.csv')
        exp.round(6).to_csv(csv_out,index=False)
        print(f'  CSV: {os.path.basename(csv_out)}')

        processed.append(dict(
            lbl=lbl, col=col, df=df, d=d, ok=ok, bad=bad,
            T_C_a=T_C_a, T_K_a=T_K_a, iT_a=iT_a, P_a=P_a, ng_a=ng_a,
            n_rxn_a=n_rxn_a, frac_d=frac_d, log10_a=log10_a,
            ln_K_a=ln_K_a, P_N2_a=P_N2_a, P_Li_a=P_Li_a,
            dH_u=dH_u, dS_u=dS_u, Teq_u=Teq_u, r2_u=r2_u,
            dH_err=dH_err, dS_err=dS_err, Teq_err=Teq_err,
            ic_u=ic_u, sl_u=sl_u, sl_log=sl_log, ic_log=ic_log,
            dH_log=dH_log, r_log=r_log,
            x_fit=x_fit, y_fit=y_fit, se_b=se_b, t95=t95,
            resid_u=resid_u, kink=kink,
            mass_ok=mass_ok, n_rxn_tot=n_rxn_tot, conv=conv,
            n_pts=len(d), single_point=False,
        ))

    THEME = dict(template='simple_white',
                 font=dict(family='Arial, sans-serif', size=12),
                 margin=dict(l=60,r=20,t=55,b=55),
                 legend=dict(bgcolor='rgba(255,255,255,0.8)',
                             bordercolor='#ddd', borderwidth=1))

    # ── Li3N reference ────────────────────────────────────────────────────
    li3n_ok=False
    if cfg['LI3N_FILE'] and os.path.isfile(cfg['LI3N_FILE']):
        df_l3=pd.read_csv(cfg['LI3N_FILE']); df_l3.columns=df_l3.columns.str.strip()
        ren={csv_h:int_k for csv_h,int_k in zip(cfg['COL_VALS'],cfg['COL_KEYS'])
             if csv_h in df_l3.columns}
        df_l3=df_l3.rename(columns=ren)
        if 'use_in_fit' not in df_l3.columns: df_l3['use_in_fit']=True
        li3n_ok=True; print(f'\nLi3N reference: {len(df_l3)} rows')

    # extrapolation using first run's kink parameters for the PT diagram
    r0=processed[0]
    T_pl_K=np.linspace(900,1450,500); T_pl_C=T_pl_K-273.15
    P_Li_ext=P_Li_sat(T_pl_K)
    T_l3=np.linspace(900,1450,300)
    P_l3=np.exp(2*(-cfg['dH_Li3N_lit']*1e3/R/T_l3+cfg['dS_Li3N_lit']/R))*P0

    print(f'\nBuilding figures ({len(processed)} run(s))...')

    # ── Fig 1: QC overview — one column per run ───────────────────────────
    n_runs = len(processed)
    fig = make_subplots(
        rows=3, cols=n_runs,
        subplot_titles=[f'{r["lbl"]}' for r in processed]*3,
        row_titles=['P_eq (MPa)', 'n_gas (mmol)', 'dn (mmol)'],
        vertical_spacing=0.10, horizontal_spacing=0.06)

    for ci, r in enumerate(processed, start=1):
        df=r['df']; ok=r['ok']; bad=r['bad']; col=r['col']
        fig.add_trace(go.Scatter(x=df.loc[ok,'T_C'],y=df.loc[ok,'P_eq_MPa'],
            mode='markers',marker=dict(color=col,size=6),name=r['lbl'],
            showlegend=(ci==1)),row=1,col=ci)
        if bad.any():
            fig.add_trace(go.Scatter(x=df.loc[bad,'T_C'],y=df.loc[bad,'P_eq_MPa'],
                mode='markers',marker=dict(color=C_RED,size=8,symbol='x'),
                showlegend=(ci==1),name='flagged'),row=1,col=ci)
        fig.add_trace(go.Scatter(x=df.loc[ok,'T_C'],y=df.loc[ok,'n_gas_mmol'],
            mode='lines+markers',marker=dict(color=col,size=5),
            line=dict(color=col),showlegend=False),row=2,col=ci)
        bar_col=[col if v else C_RED for v in ok]
        fig.add_trace(go.Bar(x=np.arange(len(df)),y=df['dn_mmol'],
            marker_color=bar_col,showlegend=False,
            error_y=dict(type='data',array=df['sigma_dn'],visible=True)),row=3,col=ci)
        fig.add_trace(go.Scatter(x=np.arange(len(df)),y=df['LOD'],mode='lines',
            line=dict(color=C_ORANGE,dash='dash',width=1.2),
            name='LOD',showlegend=(ci==1)),row=3,col=ci)
    fig.update_xaxes(title_text='T (C)',row=1); fig.update_xaxes(title_text='T (C)',row=2)
    fig.update_xaxes(title_text='Step',row=3)
    fig.update_layout(height=600, width=max(400*n_runs,700), **THEME)
    _show(fig,'fig1_qc_overview','QC overview')

    # ── Fig 2: mole tracking — one column per run ─────────────────────────
    fig = make_subplots(rows=2, cols=n_runs,
        subplot_titles=[f'{r["lbl"]}' for r in processed]*2,
        row_titles=['n (mmol)', 'Decomposition (%)'],
        vertical_spacing=0.12, horizontal_spacing=0.06)
    for ci, r in enumerate(processed, start=1):
        fig.add_trace(go.Scatter(x=r['T_C_a'],y=r['n_rxn_a'],mode='lines+markers',
            marker=dict(color=r['col'],size=6),line=dict(color=r['col']),
            name=r['lbl'],showlegend=(ci==1)),row=1,col=ci)
        if r['mass_ok']:
            fig.add_hline(y=r['n_rxn_tot'],line=dict(color=C_ORANGE,dash='dash',width=1.5),
                annotation_text=f'{r["n_rxn_tot"]:.3f} mmol',
                annotation_position='right',row=1,col=ci)
        if r['frac_d'] is not None:
            fig.add_trace(go.Scatter(x=r['T_C_a'],y=r['frac_d'],mode='lines+markers',
                marker=dict(color=r['col'],size=6),line=dict(color=r['col']),
                showlegend=False),row=2,col=ci)
            if r['mass_ok']:
                fig.add_hline(y=r['conv'],line=dict(color=C_ORANGE,dash='dash',width=1.5),
                    annotation_text=f'{r["conv"]:.1f}%',
                    annotation_position='right',row=2,col=ci)
    fig.update_xaxes(title_text='T (C)')
    fig.update_layout(height=480, width=max(420*n_runs,700), **THEME)
    _show(fig,'fig2_mole_tracking','Mole tracking')

    # ── Fig 3: van't Hoff ln K — all runs overlaid ────────────────────────
    fig = go.Figure()
    for r in processed:
        # confidence band only for single run; skip for multi to avoid clutter
        if n_runs == 1:
            fig.add_trace(_band(r['x_fit'],r['y_fit'],r['t95']*r['se_b'],
                color=f"rgba({int(r['col'][1:3],16)},{int(r['col'][3:5],16)},{int(r['col'][5:7],16)},0.12)",
                name=f"{r['lbl']} 95% CI"))
        if not r.get('single_point'):
            fig.add_trace(go.Scatter(x=r['x_fit'],y=r['y_fit'],mode='lines',
                line=dict(color=r['col'],width=2),
                name=f"{r['lbl']} fit  R2={r['r2_u']:.3f}"))
        fig.add_trace(go.Scatter(x=r['iT_a'],y=r['ln_K_a'],mode='markers',
            marker=dict(color=r['col'],size=7,
                        line=dict(color='rgba(0,0,0,0.3)',width=0.5)),
            name=r['lbl'],
            customdata=r['T_C_a'],
            hovertemplate='1/T=%{x:.6f}<br>ln K=%{y:.3f}<br>T=%{customdata:.0f} C<extra></extra>'))
    # annotation box listing all runs
    ann_lines = [f"{r['lbl']}: dH={r['dH_u']:+.1f} +/-{r['dH_err']:.1f} kJ/mol  Teq={r['Teq_u']-273.15:.0f} C"
                 for r in processed]
    ann_lines.append(f'Mode: {MODE}')
    _annot(fig, '\n'.join(ann_lines))
    fig.update_layout(height=500,width=720,
        xaxis_title='1/T (1/K)',yaxis_title='ln K',
        title="Van't Hoff -- Li3BN2 decomposition",**THEME)
    _show(fig,'fig3_vant_hoff_lnK',"Van't Hoff (ln K)")

    # ── Fig 4: Clausius-Clapeyron — all runs overlaid ─────────────────────
    fig = make_subplots(rows=1,cols=2,subplot_titles=['1/T axis','1000/T axis'])
    for ci,(xsc,xtit) in enumerate([(1.0,'1/T (1/K)'),(1000.0,'1000/T (1/K)')],start=1):
        for r in processed:
            slp=r['sl_log']/xsc
            xf=r['x_fit']*xsc; yf=slp*xf+r['ic_log']
            xd=r['iT_a']*xsc
            if not r.get('single_point'):
                fig.add_trace(go.Scatter(x=xf,y=yf,mode='lines',
                    line=dict(color=r['col'],width=2),
                    name=f"{r['lbl']} fit",showlegend=(ci==1)),row=1,col=ci)
            fig.add_trace(go.Scatter(x=xd,y=r['log10_a'],mode='markers',
                marker=dict(color=r['col'],size=7,
                            line=dict(color='rgba(0,0,0,0.3)',width=0.5)),
                name=r['lbl'],showlegend=False,
                customdata=r['T_C_a'],
                hovertemplate='x=%{x:.5f}<br>log10P=%{y:.3f}<br>T=%{customdata:.0f} C<extra></extra>'),
                row=1,col=ci)
        fig.update_xaxes(title_text=xtit,row=1,col=ci)
        fig.update_yaxes(title_text='log10 P_eq (MPa)',row=1,col=ci)
    fig.update_layout(height=460,width=1020,
        title='Clausius-Clapeyron -- Li3BN2 decomposition',**THEME)
    _show(fig,'fig4_clausius_clapeyron','Clausius-Clapeyron (log10 P)')

    # ── Fig 5: slope-kink — all runs overlaid ────────────────────────────
    fig = make_subplots(rows=1,cols=2,
        subplot_titles=['Slope-kink -- ln K','Slope-kink -- log10 P'])
    for ci,(ya_key,yg_fn,ylbl) in enumerate([
        ('ln_K_a',   lambda r: r['sl_u']*r['x_fit']+r['ic_u'],    'ln K'),
        ('log10_a',  lambda r: r['sl_log']*r['x_fit']+r['ic_log'],'log10 P'),
    ],start=1):
        for r in processed:
            ya=r[ya_key]; kink=r['kink']; idx=kink['idx']
            # global fit (dashed)
            if not r.get('single_point'):
                fig.add_trace(go.Scatter(x=r['x_fit'],y=yg_fn(r),mode='lines',
                    line=dict(color=r['col'],dash='dash',width=1.2),
                    name=f"{r['lbl']} global",showlegend=(ci==1)),row=1,col=ci)
            # segment fits and kink marker — only when enough points exist
            if not r.get('single_point') and idx > 0 and idx < len(r['iT_a']):
                x1=np.linspace(r['iT_a'][:idx].min()*0.999,r['iT_a'][:idx].max()*1.001,60)
                x2=np.linspace(r['iT_a'][idx:].min()*0.999,r['iT_a'][idx:].max()*1.001,60)
                fig.add_trace(go.Scatter(x=x1,y=kink['sl1']*x1+kink['ic1'],mode='lines',
                    line=dict(color=r['col'],width=2.5),
                    name=f"{r['lbl']} Reg.II dH1={kink['dH1']:+.0f}",
                    showlegend=(ci==1)),row=1,col=ci)
                fig.add_trace(go.Scatter(x=x2,y=kink['sl2']*x2+kink['ic2'],mode='lines',
                    line=dict(color=r['col'],width=2.5,dash='dot'),
                    name=f"{r['lbl']} Reg.III dH4={kink['dH2']:+.0f}",
                    showlegend=(ci==1)),row=1,col=ci)
                fig.add_vline(x=1/kink['T_bk'],
                    line=dict(color=r['col'],dash='dot',width=0.8),opacity=0.6,
                    row=1,col=ci)
            # data markers always shown (single points too)
            fig.add_trace(go.Scatter(x=r['iT_a'],y=ya,mode='markers',
                marker=dict(color=r['col'],size=7,
                            line=dict(color='rgba(0,0,0,0.3)',width=0.5)),
                name=r['lbl'],showlegend=False,
                customdata=r['T_C_a'],
                hovertemplate='1/T=%{x:.6f}<br>y=%{y:.3f}<br>T=%{customdata:.0f} C<extra></extra>'),
                row=1,col=ci)
        fig.update_xaxes(title_text='1/T (1/K)',row=1,col=ci)
        fig.update_yaxes(title_text=ylbl,row=1,col=ci)
    fig.update_layout(height=460,width=1020,**THEME)
    _show(fig,'fig5_slope_kink','Slope-kink (Regime II / III)')

    # ── Fig 6: dG(T) / Li mode comparison — use first run ─────────────────
    r0=processed[0]
    T_rng=np.linspace(r0['T_K_a'].min()-100,r0['T_K_a'].max()+200,500)
    li_modes=['none','saturated','stoichiometric']
    li_cols={'none':C_BLUE,'saturated':C_GREEN,'stoichiometric':C_RED}

    fig=make_subplots(rows=1,cols=2,
        subplot_titles=[f"Van't Hoff -- Li modes  ({r0['lbl']})",
                        f"dG(T) -- Li modes  ({r0['lbl']})"])
    for mode in li_modes:
        P_a0=r0['P_a']; T_K0=r0['T_K_a']; iT0=r0['iT_a']
        Pm=(P_Li_sat(T_K0) if mode=='saturated'
            else P_a0*NU_LI/(NU_LI+NU_N2) if mode=='stoichiometric' else np.zeros_like(P_a0))
        Pn=np.maximum(P_a0-Pm if mode=='saturated' else P_a0/7 if mode=='stoichiometric' else P_a0,1e-12)
        lKm=(0.5*np.log(Pn/P0) if mode in ('none','saturated')
             else 0.5*np.log(Pn/P0)+3*np.log(np.maximum(Pm/P0,1e-30)))
        dHm,dSm,Teqm,*_=vhfit(iT0,lKm)
        slm,icm,*_=stats.linregress(iT0,lKm)
        col=li_cols[mode]
        fig.add_trace(go.Scatter(x=iT0,y=lKm,mode='markers',
            marker=dict(color=col,size=5,opacity=0.6),showlegend=False),row=1,col=1)
        fig.add_trace(go.Scatter(x=r0['x_fit'],y=slm*r0['x_fit']+icm,mode='lines',
            line=dict(color=col,width=2),
            name=f'{mode}: dH={dHm:+.0f}, Teq={Teqm-273.15:.0f}C'),row=1,col=1)
        fig.add_trace(go.Scatter(x=T_rng-273.15,y=(dHm*1e3-T_rng*dSm)/1e3,mode='lines',
            line=dict(color=col,width=2),showlegend=False),row=1,col=2)
        fig.add_vline(x=Teqm-273.15,line=dict(color=col,dash='dot',width=0.8),row=1,col=2)
    fig.add_hline(y=0,line=dict(color=C_GREY,width=0.8,dash='dot'),row=1,col=2)
    fig.update_xaxes(title_text='1/T (1/K)',row=1,col=1)
    fig.update_xaxes(title_text='T (C)',row=1,col=2)
    fig.update_yaxes(title_text='ln K',row=1,col=1)
    fig.update_yaxes(title_text='dG (kJ/mol)',row=1,col=2)
    fig.update_layout(height=460,width=1020,**THEME)
    _show(fig,'fig6_dG_mode_comparison','dG(T) / Li mode comparison')

    # ── Fig 7a: P-T stability diagram — all runs overlaid ─────────────────
    fig=make_subplots(rows=1,cols=2,
        subplot_titles=['P-T stability diagram (linear)','P-T stability diagram (log10)'])
    for ci,(yfn,ylbl) in enumerate([
        (lambda p:p, 'P_eq (MPa)'),
        (lambda p:np.log10(np.maximum(p,1e-9)), 'log10 P_eq (MPa)'),
    ],start=1):
        # each run's data + extrapolation
        for r in processed:
            kink=r['kink']
            P_N2_ext=np.exp(2*(kink['sl1']*(1/T_pl_K)+kink['ic1']))*P0
            P_tot_ext=P_N2_ext+P_Li_ext
            fig.add_trace(go.Scatter(x=r['T_C_a'],y=yfn(r['P_a']),mode='markers',
                marker=dict(color=r['col'],size=7,
                            line=dict(color='rgba(0,0,0,0.3)',width=0.5)),
                name=r['lbl'],
                customdata=r['T_K_a'],
                hovertemplate='T=%{x:.0f} C<br>P=%{y:.4f}<br>T_K=%{customdata:.0f} K<extra></extra>'),
                row=1,col=ci)
            fig.add_trace(go.Scatter(x=T_pl_C,y=yfn(P_tot_ext),mode='lines',
                line=dict(color=r['col'],width=1.8),
                name=f'{r["lbl"]} equil.',showlegend=(ci==1)),row=1,col=ci)
        # Li3N reference
        fig.add_trace(go.Scatter(x=T_l3-273.15,y=yfn(P_l3),mode='lines',
            line=dict(color='#888',width=1.8,dash='dash'),
            name='Li3N equil. (Metzger et al.)'),row=1,col=ci)
        if li3n_ok:
            dl=df_l3[df_l3['use_in_fit']] if 'use_in_fit' in df_l3.columns else df_l3
            fig.add_trace(go.Scatter(x=dl['T_C'],y=yfn(dl['P_eq_MPa'].values),
                mode='markers',marker=dict(color='#888',size=7,symbol='triangle-up'),
                name='Li3N measured'),row=1,col=ci)
        fig.add_trace(go.Scatter(x=T_pl_C,y=yfn(P_Li_ext),mode='lines',
            line=dict(color=C_ORANGE,dash='dot',width=1.2),
            name='P_Li_sat'),row=1,col=ci)
        for Tv,lb,cv in [(813,'Tm(Li3N)',C_RED),(863,'',C_GREY),(916,'Tm(Li3BN2)',C_GREY)]:
            fig.add_vline(x=Tv,line=dict(color=cv,dash='dot',width=0.7),opacity=0.5,
                annotation_text=lb,annotation_position='top right',row=1,col=ci)
        if ci==1:
            fig.add_hline(y=50,line=dict(color=C_GREY,dash='dash',width=0.7),
                annotation_text='~50 MPa (Maestre 2024)',annotation_position='right',
                row=1,col=ci)
        fig.update_xaxes(title_text='T (C)',range=[750,T_pl_C.max()],row=1,col=ci)
        fig.update_yaxes(title_text=ylbl,row=1,col=ci)
    fig.update_layout(height=490,width=1120,**THEME)
    _show(fig,'fig7a_PT_phase_diagram','P-T stability diagram')

    # ── Fig 7b: Li3BN2 vs Li3N comparison — all runs ──────────────────────
    dH_l3n_lit=cfg['dH_Li3N_lit']; dS_l3n_lit=cfg['dS_Li3N_lit']
    # use x_fit from first run as common grid
    xf0=processed[0]['x_fit']
    ln_K_l3n_fit=-dH_l3n_lit*1e3/R*xf0+dS_l3n_lit/R
    log10_P_l3n_fit=np.log10(np.maximum(np.exp(2*ln_K_l3n_fit)*P0,1e-12))

    fig=make_subplots(rows=1,cols=2,
        subplot_titles=['ln K vs. 1/T  (Li3BN2 and Li3N)',
                        'log10 P vs. 1/T  (Li3BN2 and Li3N)'])
    for ci,(ya_key,yfit_l3n,ytit) in enumerate([
        ('ln_K_a',  ln_K_l3n_fit,    'ln K'),
        ('log10_a', log10_P_l3n_fit, 'log10 P (MPa)'),
    ],start=1):
        for r in processed:
            fig.add_trace(go.Scatter(x=r['iT_a'],y=r[ya_key],mode='markers',
                marker=dict(color=r['col'],size=7,
                            line=dict(color='rgba(0,0,0,0.3)',width=0.5)),
                name=r['lbl'],showlegend=(ci==1),
                customdata=r['T_C_a'],
                hovertemplate='1/T=%{x:.6f}<br>y=%{y:.3f}<br>T=%{customdata:.0f} C<extra></extra>'),
                row=1,col=ci)
            if not r.get('single_point'):
                yfit_run=(r['y_fit'] if ya_key=='ln_K_a' else r['sl_log']*r['x_fit']+r['ic_log'])
                fig.add_trace(go.Scatter(x=r['x_fit'],y=yfit_run,mode='lines',
                    line=dict(color=r['col'],width=1.8),
                    name=f'{r["lbl"]} fit',showlegend=(ci==1)),row=1,col=ci)
        # Li3N reference line
        fig.add_trace(go.Scatter(x=xf0,y=yfit_l3n,mode='lines',
            line=dict(color='#888',width=1.8,dash='dash'),
            name='Li3N (Metzger et al.)'),row=1,col=ci)
        if li3n_ok:
            dl=df_l3[df_l3['use_in_fit']] if 'use_in_fit' in df_l3.columns else df_l3
            if 'inv_T' not in dl.columns: dl=dl.copy(); dl['inv_T']=1/dl['T_K']
            y_l3d=(0.5*np.log(dl['P_eq_MPa'].values/P0) if ci==1
                   else np.log10(dl['P_eq_MPa'].values))
            fig.add_trace(go.Scatter(x=dl['inv_T'],y=y_l3d,mode='markers',
                marker=dict(color='#888',size=7,symbol='triangle-up'),
                name='Li3N data',showlegend=(ci==1)),row=1,col=ci)
        if ci==1:
            ann='\n'.join(
                [f'{r["lbl"]}: dH={r["dH_u"]:+.1f} kJ/mol' for r in processed]
                + [f'Li3N: dH={dH_l3n_lit:+.1f} kJ/mol',
                   f'dH3 ~ {processed[0]["dH_u"]-dH_l3n_lit:+.1f} kJ/mol  (lit +117.5)'])
            _annot(fig,ann,xref='paper',yref='paper',x=0.46,y=0.05)
        fig.update_xaxes(title_text='1/T (1/K)',row=1,col=ci)
        fig.update_yaxes(title_text=ytit,row=1,col=ci)
    fig.update_layout(height=480,width=1050,
        title='Li3BN2 vs. Li3N equilibrium comparison  (Hess K3)',**THEME)
    _show(fig,'fig7b_Li3BN2_vs_Li3N_comparison','Li3BN2 vs. Li3N comparison')

    # ── Fig 8: residuals — all runs overlaid ──────────────────────────────
    fig=make_subplots(rows=1,cols=2,
        subplot_titles=['Residuals -- ln K','Residuals -- log10 P'])
    for ci,ya_key in enumerate(['ln_K_a','log10_a'],start=1):
        for r in processed:
            if ci==1:
                fit_y=r['sl_u']*r['iT_a']+r['ic_u']
            else:
                fit_y=r['sl_log']*r['iT_a']+r['ic_log']
            ra=r[ya_key]-fit_y
            rmse=np.sqrt((ra**2).mean())
            fig.add_trace(go.Scatter(x=r['T_C_a'],y=ra,mode='markers',
                marker=dict(color=r['col'],size=7,
                            line=dict(color='rgba(0,0,0,0.3)',width=0.5)),
                name=f'{r["lbl"]} (RMSE={rmse:.4f})',showlegend=(ci==1),
                hovertemplate='T=%{x:.0f} C<br>res=%{y:.4f}<extra></extra>'),
                row=1,col=ci)
        fig.add_hline(y=0,line=dict(color=C_GREY,dash='dot',width=0.8),row=1,col=ci)
        lb='ln K' if ci==1 else 'log10 P'
        fig.update_xaxes(title_text='T (C)',row=1,col=ci)
        fig.update_yaxes(title_text=f'Residual ({lb})',row=1,col=ci)
    fig.update_layout(height=380,width=920,**THEME)
    _show(fig,'fig8_residuals','Residuals')

    # ── summary ───────────────────────────────────────────────────────────
    print('\n'+'='*56+'\n  RESULTS SUMMARY\n'+'='*56)
    for r in processed:
        kink=r['kink']
        print(f'\n  [{r["lbl"]}]')
        if r.get('single_point'):
            print(f'    Single point: T={r["T_C_a"][0]:.1f} C  P={r["P_a"][0]:.4f} MPa  log10P={r["log10_a"][0]:.3f}')
            continue
        print(f'    dH = {r["dH_u"]:+.2f} +/- {r["dH_err"]:.2f} kJ/mol')
        print(f'    dS = {r["dS_u"]:+.2f} +/- {r["dS_err"]:.2f} J/(mol K)')
        print(f'    T_eq = {r["Teq_u"]-273.15:.0f} +/- {r["Teq_err"]:.0f} C  R2={r["r2_u"]:.5f}')
        print(f'    Kink {kink["T_bk"]-273.15:.0f} C: dH1={kink["dH1"]:+.1f}  dH4={kink["dH2"]:+.1f}  dH3={kink["dH3"]:+.1f} kJ/mol')
        if r['mass_ok']:
            print(f'    Conversion: {r["conv"]:.2f}%  n(rxn): {r["n_rxn_tot"]:.4f} mmol')
    print('='*56)
    print(f'  {len(_figures)} plots ready -- click Save plots.')


# ════════════════════════════════════════════════════════════════════════
#  BUTTON CALLBACKS
# ════════════════════════════════════════════════════════════════════════
def on_run(b):
    cfg=collect_config()
    with out_text: clear_output(wait=True)
    with out_btns: clear_output(wait=True)
    with out_plots: clear_output(wait=True)
    if not cfg['RUNS']:
        out_text.append_stdout('Please add at least one CSV file in the Data & masses tab.\n'); return
    missing=[r['DATA_FILE'] for r in cfg['RUNS'] if not os.path.isfile(r['DATA_FILE'])]
    if missing:
        out_text.append_stdout('File(s) not found:\n' + '\n'.join(f'  {p}' for p in missing) + '\n'); return

    _figures.clear()
    btn_save.disabled=True; btn_save.description='Save plots'
    btn_run.disabled=True;  btn_run.description='Running...'
    status.value=f'<span style="font-size:12px;color:{C_ORANGE}">Running...</span>'

    def _worker():
        tee=_Tee(sys.stdout); sys.stdout=tee
        try:
            run_analysis(cfg)
            OUT=cfg['OUT_DIR']; os.makedirs(OUT,exist_ok=True)
            ts=datetime.now().strftime('%y%m%d_%H%M%S')
            log_p=os.path.join(OUT,f'run_log_{ts}.txt')
            with open(log_p,'w',encoding='utf-8') as lf: lf.write(tee.getvalue())
            out_text.append_stdout(f'  Log: run_log_{ts}.txt\n')
            n=len(_figures)
            btn_save.disabled=False; btn_save.description=f'Save {n} plots'
            status.value=(f'<span style="font-size:12px;color:{C_GREEN}">'
                          f'Done -- {n} plots ready. Click Save plots.</span>')
        except Exception:
            tb=traceback.format_exc()
            out_text.append_stdout(f'\n-- Error --\n{tb}\n')
            status.value=f'<span style="font-size:12px;color:{C_RED}">Error -- see output above.</span>'
        finally:
            sys.stdout=tee._orig; btn_run.disabled=False; btn_run.description='Run analysis'

    threading.Thread(target=_worker,daemon=True).start()


def on_save(b):
    if not _figures:
        out_text.append_stdout('No plots to save -- run analysis first.\n'); return
    OUT=v_out_dir.value.strip(); os.makedirs(OUT,exist_ok=True)
    ts=datetime.now().strftime('%y%m%d_%H%M%S')
    btn_save.disabled=True; btn_save.description='Saving...'
    def _save():
        try:
            out_text.append_stdout(f'\nSaving {len(_figures)} plots [{ts}]...\n')
            for fig,base in _figures:
                fname=f'{base}_{ts}.png'
                fig.write_image(os.path.join(OUT,fname),scale=2)
                out_text.append_stdout(f'  {fname}\n')
            out_text.append_stdout(f'Done. Saved to: {OUT}\n')
            status.value=f'<span style="font-size:12px;color:{C_GREEN}">Saved to {OUT}</span>'
        except Exception:
            out_text.append_stdout(f'\n-- Save error --\n{traceback.format_exc()}' + "\n")
        finally:
            btn_save.disabled=False; btn_save.description=f'Save {len(_figures)} plots'
    threading.Thread(target=_save,daemon=True).start()


def on_clear(b):
    with out_text: clear_output()
    with out_btns: clear_output()
    with out_plots: clear_output()
    _figures.clear()
    btn_save.disabled=True; btn_save.description='Save plots'
    status.value='<span style="font-size:12px;color:#888">Ready.</span>'


btn_run.on_click(on_run)
btn_save.on_click(on_save)
btn_clear.on_click(on_clear)

print('UI ready.')
```
